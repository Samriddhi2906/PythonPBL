```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_selection import SelectKBest,chi2
import pandas as pd
```

```python
# LOAD DATASET                  
df = pd.read_csv("final_clean_dataset.csv")
df['clean_review'] = df['clean_review'].fillna('')
df['clean_summary'] = df['clean_summary'].fillna('')
```

```python
# ================================
# COMBINE REVIEW + SUMMARY
# ================================
df['combined_text'] = df['clean_review'] + " " + df['clean_summary']
```

```python
# TF-IDF (clean_review)
tfidf_review = TfidfVectorizer(max_features=1000)
X_review = tfidf_review.fit_transform(df['clean_review'])
# TF-IDF (clean_summary)
tfidf_summary = TfidfVectorizer(max_features=1000)
X_summary = tfidf_summary.fit_transform(df['clean_summary'])
```
